```python
%sql
-- Create the top-level catalog for the Blue Ridge Parkway closure project.
-- All medallion schemas (bronze, silver, gold) live inside this catalog.

CREATE CATALOG IF NOT EXISTS brp
COMMENT 'Blue Ridge Parkway closure Project';
```

```python
%sql
-- Create the medallion architecture schemas (bronze, silver, gold) in the brp catalog.

CREATE SCHEMA IF NOT EXISTS brp.bronze 
COMMENT 'Raw data as exactly as received from source';
CREATE SCHEMA IF NOT EXISTS brp.silver 
COMMENT 'Cleaned standardized data';
CREATE SCHEMA IF NOT EXISTS brp.gold 
COMMENT 'Aggregated dataset for reporting and dashboards'
```

```python
%sql
-- Bronze layer volume: stores the daily raw downloads (road status HTML and milepost GeoJSON) before any parsing or cleaning happens in the silver layer.

CREATE VOLUME IF NOT EXISTS brp.bronze.raw
COMMENT 'Daily raw downloads: road status HTML and milepost GeoJSON';
```

```python
%sql
-- Lists all schemas currently in the brp catalog (should show bronze, silver, and gold, as well as build in default and information_schema).

SHOW SCHEMAS IN brp;
```
