```python
import requests
```

```python
# Define the list of sites to test for network reachability from this workspace.
# Each entry pairs a human-readable name with a URL
# The loop below iterates over these and reports REACHED (with HTTP status code) or BLOCKED (with the exception) for each request.

# --- Configuration: sites to test (name -> URL) ---
sites = {
    # NPS Blue Ridge Parkway road closure status page
    "NPS road status page": "https://www.nps.gov/blri/planyourvisit/roadclosures.htm", 

    # ArcGIS FeatureServer metadata for parkway mileposts
    "NPS ArcGIS mileposts": "https://services1.arcgis.com/fBc8EJBxQRMcHlei/arcgis/rest/services/SEAC_BLRI_Mileposts/FeatureServer",

    # Neutral control site to confirm general outbound internet access works
    "GitHub": "https://raw.githubusercontent.com/github/gitignore/main/Python.gitignore",
}

# --- Probe each site and report the result ---
for name, url in sites.items():
    try:
        # Attempt the request with a timeout so unreachable sites fail fast
        response = requests.get(url, timeout=15)
        print(f"REACHED  {response.status_code}  {name}")
    except Exception as error:
        # Any network-level failure (DNS, timeout, SSL, etc.) lands here
        print(f"BLOCKED  {name}  ({type(error).__name__}: {error})")
```
